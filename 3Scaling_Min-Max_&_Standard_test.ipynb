```python
import math
import statistics
import numpy as np
import pandas as pd
```

```python
#1

num = np.array([18,42,66,90])

min_ = min(num)
max_ = max(num)
maxmin = max_ - min_

scaled_num = (num - min_) / maxmin
print("original", num)
print("scaled", np.round(scaled_num,3))
```

```python
#2

num = np.array([8, 10, 10, 14, 14, 16])

mean_ = sum(num) / len(num)
deviations = num - mean_
squared = deviations ** 2
variance = squared.sum() / len(num)
std_ = np.sqrt(variance)

print("mean:", mean_)
print("deviations:", deviations)
print("std:", round(std_, 3))
```
