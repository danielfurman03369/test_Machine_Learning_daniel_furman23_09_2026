```python
#4


x = np.array([1, 3, 5, 7])
y = np.array([19, 25, 41, 51])

#a

pred_9 = 6 * 9 + 10
print("a:", pred_9)
print("***************************************")

#b

pred_5 = 6 * 5 + 10
resid5 = 41 - pred_5
print("b:", resid5)
print("***************************************")

#c

y_hat = 6 * x + 10
e = y - y_hat
e_sq = e ** 2

sse = np.sum(e_sq)
mse = sse / len(y)
rmse = np.sqrt(mse)

print("c:")
print("y_hat:", y_hat)
print("e:", e)
print("e^2:", e_sq)
print("sse:", sse)
print("mse:", mse)
print("rmse:", round(rmse, 3))
print("***************************************")

#d
#about 96.9% of the variation in quiz scores is explained by hours studied,
#so the line fits the data very well
mean_y = np.sum(y) / len(y)
dev_y = y - mean_y
dev_y_sq = dev_y ** 2
sst = np.sum(dev_y_sq)
r2 = 1 - sse / sst
print("d:")
print("mean of y:", mean_y)
print("sst:", sst)
print("r^2:", round(r2, 3))
```
