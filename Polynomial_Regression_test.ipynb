```python
#5

r2 = 0.94
n = 10
p = 3

a = 1 - r2
b = (n - 1) / (n - p - 1)
adj_r2 = 1 - a * b

print("1 - r^2:", round(a, 3))
print("(n - 1) / (n - p - 1):", b)
print("adj_r^2:", round(adj_r2, 3))
```
