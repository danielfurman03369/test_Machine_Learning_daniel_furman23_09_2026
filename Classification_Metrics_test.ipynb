```python
#10

#a
tp = 45
fp = 60 - 45
fn = 25
tn = 190 - 25
tot = 250

#b
acc = (tp + tn) / tot
prec = tp / (tp + fp)
recall = tp / (tp + fn)
f1 = 2 * prec * recall / (prec + recall)

print("a:")
print("tp:", tp, "| fp:", fp, "| fn:", fn, "| tn:", tn)
print("***************************************")
print("b:")
print("accuracy:", round(acc, 3))
print("precision:", round(prec, 3))
print("recall:", round(recall, 3))
print("f1:", round(f1, 3))
```
